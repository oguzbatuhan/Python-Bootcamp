## continue - break -pass

```python
my_list = [10, 20, 30, 40 ,50, 60]
```

```python
for numbar in my_list:
    print(numbar)
    if numbar == 40:
        print("yes")
```

```python
for numbar in my_list:
    print(numbar)
    if numbar == 40:
        print("yes")
        break
```

```python
for numbar in my_list:
    if numbar == 40:
        continue
    print(numbar)
```

```python
for numbar in my_list:
    pass
```
